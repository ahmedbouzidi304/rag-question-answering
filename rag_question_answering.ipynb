**Why RAG?** LLMs answer fluently but can hallucinate and know nothing about private or recent documents. Retrieval-Augmented Generation first **retrieves** the most relevant passages from a knowledge base, then asks the LLM to answer **grounded in those passages**, with sources.

**What this notebook delivers**
- a searchable knowledge base of Wikipedia paragraphs with gold question → passage labels;
- four retrievers, benchmarked with **Recall@k**, **MRR** and latency;
- an open LLM answering with and without retrieval, scored with the official SQuAD **EM / F1** definitions;
- an error analysis separating *retrieval failures* from *generation failures*, and a reusable `ask()` function.

The SQuAD validation split contains questions written by annotators about Wikipedia paragraphs. De-duplicating the paragraphs gives a **document collection**, and each question keeps a pointer to its **gold passage**: a labelled retrieval benchmark for free.

```python
squad = load_dataset("rajpurkar/squad", split="validation")
qa = squad.to_pandas()
qa["answers"] = qa["answers"].apply(lambda a: list(a["text"]))

corpus = qa[["title", "context"]].drop_duplicates("context").reset_index(drop=True)
corpus["title"] = corpus["title"].str.replace("_", " ")
corpus["doc_id"] = np.arange(len(corpus))
corpus["text"] = corpus["title"] + ". " + corpus["context"]   # title gives the passage its topic
qa["gold_doc"] = qa["context"].map(dict(zip(corpus["context"], corpus["doc_id"])))

print(f"{len(qa):,} questions | {len(corpus):,} passages | {corpus['title'].nunique()} Wikipedia articles")
corpus.head(3)
```

Two benchmarks with the same seeded sample:
- **retrieval**: 1,000 questions, *did the gold passage come back, and how high?*
- **generation**: the first 300 of those, *is the final answer correct?* (LLM inference is the expensive step).

One caveat is stated up front: SQuAD questions were written by people *looking at* the passage, so they share many words with it. This favours lexical search (BM25) more than real user queries would.

```python
eval_qa = qa.sample(N_RETRIEVAL_EVAL, random_state=SEED).reset_index(drop=True)
gen_qa = eval_qa.iloc[:N_GEN_EVAL].reset_index(drop=True)
gold = eval_qa["gold_doc"].to_numpy()
print(f"retrieval eval: {len(eval_qa)} questions over {eval_qa['title'].nunique()} articles")
eval_qa[["title", "question", "answers"]].head(5)
```

```python
def evaluate_retrieval(ranked, gold, ks=(1, 3, 5, 10)) -> dict:
    """Recall@k = gold passage in top-k; MRR@10 = mean of 1/rank of the gold passage."""
    ranked = np.array([list(r)[:max(ks)] for r in ranked])
    hits = ranked == np.asarray(gold)[:, None]
    out = {f"Recall@{k}": hits[:, :k].any(axis=1).mean() for k in ks}
    first = np.where(hits.any(axis=1), hits.argmax(axis=1) + 1, np.inf)
    out["MRR@10"] = float(np.mean(1.0 / first))
    return out

retrieval_runs, latency_ms = {}, {}
```

BM25 scores passages by term overlap with the query, weighted by how rare each term is (IDF) and normalised for passage length. It is fast, needs no training, and is still the baseline every retriever must beat.

```python
STOPWORDS = set("""a an the of in on at to for from by with and or is are was were be been being what which who
whom whose when where why how did do does that this these those it its as into than""".split())

def tokenize(text: str) -> list[str]:
    return [t for t in re.findall(r"[a-z0-9]+", text.lower()) if t not in STOPWORDS]

bm25 = BM25Okapi([tokenize(t) for t in corpus["text"]])

def bm25_search(query: str, k: int = POOL) -> list[int]:
    scores = bm25.get_scores(tokenize(query))
    return np.argsort(-scores)[:k].tolist()

t0 = time.time()
bm25_ranked = [bm25_search(q) for q in eval_qa["question"]]
latency_ms["BM25"] = (time.time() - t0) / len(eval_qa) * 1000
retrieval_runs["BM25"] = bm25_ranked
evaluate_retrieval(bm25_ranked, gold)
```

A bi-encoder maps queries and passages into the same vector space; relevance is the cosine similarity. Passages are embedded **once, offline**, and stored in a FAISS inner-product index (vectors are L2-normalised, so inner product = cosine). BGE models expect an instruction prefix on queries only.

```python
embedder = SentenceTransformer(EMBED_ID, device=DEVICE)
QUERY_PREFIX = "Represent this sentence for searching relevant passages: "

n_tokens = [len(ids) for ids in embedder.tokenizer(corpus["text"].tolist())["input_ids"]]
print(f"passages longer than 512 tokens: {np.mean(np.array(n_tokens) > 512):.1%}")

t0 = time.time()
doc_emb = embedder.encode(corpus["text"].tolist(), batch_size=128, normalize_embeddings=True,
                          convert_to_numpy=True, show_progress_bar=True).astype("float32")
index = faiss.IndexFlatIP(doc_emb.shape[1])
index.add(doc_emb)
print(f"indexed {index.ntotal:,} passages x {doc_emb.shape[1]} dims in {time.time() - t0:.1f}s")

def dense_search(queries: list[str], k: int = POOL) -> np.ndarray:
    q = embedder.encode([QUERY_PREFIX + q for q in queries], batch_size=128,
                        normalize_embeddings=True, convert_to_numpy=True).astype("float32")
    return index.search(q, k)[1]

t0 = time.time()
dense_ranked = dense_search(eval_qa["question"].tolist())
latency_ms["Dense"] = (time.time() - t0) / len(eval_qa) * 1000
retrieval_runs["Dense"] = dense_ranked.tolist()
evaluate_retrieval(dense_ranked, gold)
```

BM25 and dense search fail on *different* questions (exact names and numbers vs. paraphrases). **RRF** merges the two rankings using only ranks, so no score calibration is needed:

$$\text{RRF}(d) = \sum_{r \in \text{rankers}} \frac{1}{60 + \text{rank}_r(d)}$$

```python
def rrf(*rankings, k: int = 60, top: int = POOL) -> list[int]:
    scores = collections.defaultdict(float)
    for ranking in rankings:
        for rank, doc in enumerate(ranking):
            scores[doc] += 1.0 / (k + rank + 1)
    return [doc for doc, _ in sorted(scores.items(), key=lambda kv: -kv[1])[:top]]

t0 = time.time()
hybrid_ranked = [rrf(b, d) for b, d in zip(bm25_ranked, dense_ranked.tolist())]
latency_ms["Hybrid (RRF)"] = latency_ms["BM25"] + latency_ms["Dense"] + (time.time() - t0) / len(eval_qa) * 1000
retrieval_runs["Hybrid (RRF)"] = hybrid_ranked
evaluate_retrieval(hybrid_ranked, gold)
```

A cross-encoder feeds *(query, passage)* jointly through a transformer, which is far more accurate than comparing two independent vectors but too slow to run over the whole corpus. The standard pattern is **retrieve-then-rerank**: re-score only the top-20 hybrid candidates.

```python
reranker = CrossEncoder(RERANK_ID, device=DEVICE, max_length=512)
RERANK_POOL = 20

def rerank(query: str, candidates: list[int], top: int = TOP_K) -> list[int]:
    scores = reranker.predict([(query, corpus["text"].iat[d]) for d in candidates], show_progress_bar=False)
    return [candidates[j] for j in np.argsort(-scores)[:top]]

t0 = time.time()
pairs = [(q, corpus["text"].iat[d]) for q, c in zip(eval_qa["question"], hybrid_ranked) for d in c[:RERANK_POOL]]
pair_scores = reranker.predict(pairs, batch_size=128, show_progress_bar=True).reshape(len(eval_qa), RERANK_POOL)
reranked = [[c[j] for j in np.argsort(-s)] for c, s in zip(hybrid_ranked, pair_scores)]
latency_ms["Hybrid + rerank"] = latency_ms["Hybrid (RRF)"] + (time.time() - t0) / len(eval_qa) * 1000
retrieval_runs["Hybrid + rerank"] = reranked
evaluate_retrieval(reranked, gold)
```

The prompt asks for the **shortest exact answer span** and to reply `unknown` when the context does not contain the answer, which limits hallucination and makes answers comparable with SQuAD's gold spans. Decoding is greedy (deterministic) and batched with left padding.

```python
tok = AutoTokenizer.from_pretrained(LLM_ID, padding_side="left")
llm = AutoModelForCausalLM.from_pretrained(
    LLM_ID, torch_dtype=torch.float16 if DEVICE == "cuda" else torch.float32).to(DEVICE).eval()
print(f"{LLM_ID}: {sum(p.numel() for p in llm.parameters()) / 1e9:.2f}B parameters")

SYSTEM_RAG = ("You are a precise extractive question-answering assistant. Answer with the shortest exact span "
              "copied from the context (usually 1-5 words). Do not write a full sentence. "
              "If the context does not contain the answer, reply exactly: unknown")
SYSTEM_CLOSED = "You are a concise question-answering assistant. Answer with the shortest possible answer (1-5 words)."

def build_messages(question: str, passages: list[str] | None = None) -> list[dict]:
    if passages is None:
        return [{"role": "system", "content": SYSTEM_CLOSED},
                {"role": "user", "content": f"Question: {question}\nAnswer:"}]
    context = "\n\n".join(f"[{i + 1}] {p}" for i, p in enumerate(passages))
    return [{"role": "system", "content": SYSTEM_RAG},
            {"role": "user", "content": f"Context:\n{context}\n\nQuestion: {question}\nAnswer:"}]

@torch.no_grad()
def generate(conversations: list[list[dict]], batch_size: int = 16, max_new_tokens: int = 24) -> list[str]:
    answers = []
    for i in range(0, len(conversations), batch_size):
        prompts = [tok.apply_chat_template(c, tokenize=False, add_generation_prompt=True)
                   for c in conversations[i:i + batch_size]]
        enc = tok(prompts, return_tensors="pt", padding=True).to(DEVICE)
        out = llm.generate(**enc, max_new_tokens=max_new_tokens, do_sample=False,
                           temperature=None, top_p=None, top_k=None, pad_token_id=tok.pad_token_id)
        answers += tok.batch_decode(out[:, enc["input_ids"].shape[1]:], skip_special_tokens=True)
    return [a.strip().split("\n")[0].strip().rstrip(".") for a in answers]

# smoke test on one question with its gold passage
q0 = gen_qa.iloc[0]
print("Q:", q0["question"], "| gold:", q0["answers"][0])
print("A:", generate([build_messages(q0["question"], [corpus["text"].iat[q0["gold_doc"]]])])[0])
```

**Exact Match** = normalised prediction equals a gold answer. **F1** = token overlap with the best-matching gold answer. Normalisation lower-cases and removes punctuation and articles.

```python
def normalize(s: str) -> str:
    s = "".join(ch for ch in s.lower() if ch not in set(string.punctuation))
    s = re.sub(r"\b(a|an|the)\b", " ", s)
    return " ".join(s.split())

def f1_score(pred: str, gold: str) -> float:
    p, g = normalize(pred).split(), normalize(gold).split()
    common = sum((collections.Counter(p) & collections.Counter(g)).values())
    if common == 0:
        return 0.0
    precision, recall = common / len(p), common / len(g)
    return 2 * precision * recall / (precision + recall)

def score(preds: list[str], golds: list[list[str]]) -> tuple[np.ndarray, np.ndarray]:
    em = np.array([max(normalize(p) == normalize(a) for a in ans) for p, ans in zip(preds, golds)], dtype=float)
    f1 = np.array([max(f1_score(p, a) for a in ans) for p, ans in zip(preds, golds)])
    return em, f1

score(["Denver Broncos", "the broncos"], [["Denver Broncos"], ["Denver Broncos"]])
```

Four settings on the same 300 questions:

| Setting | Context given to the LLM |
|---|---|
| Closed-book | none: the model's own memory |
| RAG · Dense | top-3 dense passages |
| RAG · best retriever | top-3 passages from the winner of the benchmark |
| Oracle | the gold passage: an upper bound for the generator alone |

```python
questions, golds = gen_qa["question"].tolist(), gen_qa["answers"].tolist()
texts = corpus["text"].to_numpy()
ctx_ids = {
    "RAG · Dense top-3": [r[:K_CONTEXT] for r in retrieval_runs["Dense"][:N_GEN_EVAL]],
    f"RAG · {BEST_RETRIEVER} top-3": [list(r)[:K_CONTEXT] for r in retrieval_runs[BEST_RETRIEVER][:N_GEN_EVAL]],
    "Oracle (gold passage)": [[g] for g in gen_qa["gold_doc"]],
}
settings = {"Closed-book (no retrieval)": [build_messages(q) for q in questions]}
settings |= {name: [build_messages(q, list(texts[ids])) for q, ids in zip(questions, c)] for name, c in ctx_ids.items()}

gen_rows, predictions = [], {}
for name, convs in settings.items():
    t0 = time.time()
    preds = generate(convs)
    em, f1 = score(preds, golds)
    predictions[name] = preds
    hit = np.mean([g in ids for g, ids in zip(gen_qa["gold_doc"], ctx_ids[name])]) if name in ctx_ids else np.nan
    gen_rows.append({"setting": name, "Exact Match": em.mean(), "F1": f1.mean(), "gold passage in context": hit,
                     "'unknown' rate": np.mean([p.lower() == "unknown" for p in preds]),
                     "s / question": (time.time() - t0) / len(convs)})
    print(f"{name:<32} EM={em.mean():.3f}  F1={f1.mean():.3f}")

gen_results = pd.DataFrame(gen_rows).set_index("setting")
gen_results.to_csv(RESULTS / "generation_benchmark.csv")
gen_results.style.format("{:.3f}", na_rep="—")
```

`ask()` chains the winning retriever and the LLM for a single question and prints the cited sources. The last question is deliberately outside the knowledge base, to check that the system abstains instead of hallucinating.

```python
def retrieve(query: str, k: int = K_CONTEXT, method: str = BEST_RETRIEVER) -> list[int]:
    bm, de = bm25_search(query), dense_search([query])[0].tolist()
    if method == "BM25":
        return bm[:k]
    if method == "Dense":
        return de[:k]
    fused = rrf(bm, de)
    return fused[:k] if method == "Hybrid (RRF)" else rerank(query, fused[:RERANK_POOL], top=k)

def ask(question: str, k: int = K_CONTEXT) -> str:
    t0 = time.time()
    ids = retrieve(question, k)
    answer = generate([build_messages(question, list(texts[ids]))])[0]
    print(f"Q: {question}\nA: {answer}    ({(time.time() - t0) * 1000:.0f} ms)")
    for rank, i in enumerate(ids, 1):
        print(f"   [{rank}] {corpus['title'].iat[i]}: {corpus['context'].iat[i][:100]}…")
    print()
    return answer

for q in ["Which team won Super Bowl 50?",
          "What is the largest city in Poland?",
          "Who did Nikola Tesla work for when he arrived in New York?",
          "What is the boiling point of mercury on Mars?"]:
    ask(q)
```
